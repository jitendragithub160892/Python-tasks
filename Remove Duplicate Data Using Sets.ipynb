```python
numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]
```

# Without using loops

```python
# Print the original list

print(numbers)

# Convert the list into a set

set1=set(numbers)
print(set1)

# Convert the set back into a list

print(list(set1))

# Print the number of original elements

print(len(numbers))

# Print the number of unique elements
print(len(set(numbers)))

# Create another example using duplicate student names.

students=["Jitu","Ayush","Piyush","Alok","Awanish","Jitu","Ayush"]

print(set(students))

```

# Observe which duplicates disappear

10,20 and 30 duplicates disappear

Set is useful as it removes the duplicate data
