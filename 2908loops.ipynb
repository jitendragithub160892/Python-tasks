```python
l=[4,5,6,89,3,5,7,8]

l_even=[]
l_odd=[]

for i in l:
    if i %2==0:
        l_even.append(i)
    else:
        l_odd.append(i)

print(l_even)
print(l_odd)
```

```python
factorial=1
n=int(input("Enter your number to find factorial: "))

for i in range(1,n+1):
    print(i)
    factorial=factorial*i

print("factorial of ",n ," is ",factorial)
```
