```python
# list

l1 = [1, 2, 3]
print(type(l1))

l2 = ["Python", "Java", "C++"]
print(type(l2))

l3 = [10, 20, 30, 40]
print(type(l3))

l4 = []
print(type(l4))

l5 = [1, "Python", 3.14]
print(type(l5))

l6 = [True, False, True]
print(type(l6))

l7 = ["Data", "Science"]
print(type(l7))

l8 = [100]
print(type(l8))

l9 = [[1, 2], [3, 4]]
print(type(l9))

l10 = [10.5, 20.5, 30.5]
print(type(l10))

l11 = ["A", "B", "C"]
print(type(l11))

l12 = [True, 10, "Python"]
print(type(l12))

l13 = [1, 1, 2, 2, 3]
print(type(l13))

l14 = [-1, -2, -3]
print(type(l14))

l15 = ["Apple", "Banana", "Mango"]
print(type(l15))
```

```python
# tuple

t1 = (1, 2, 3)
print(type(t1))

t2 = ("Python", "Java", "C++")
print(type(t2))

t3 = (10, 20, 30)
print(type(t3))

t4 = ()
print(type(t4))

t5 = (1, "Python", 3.14)
print(type(t5))

t6 = (True, False)
print(type(t6))

t7 = ("Data", "Science")
print(type(t7))

t8 = (100,)
print(type(t8))

t9 = ((1, 2), (3, 4))
print(type(t9))

t10 = (10.5, 20.5)
print(type(t10))

t11 = ("A", "B", "C")
print(type(t11))

t12 = (True, 10, "Python")
print(type(t12))

t13 = (1, 1, 2, 2, 3)
print(type(t13))

t14 = (-1, -2, -3)
print(type(t14))

t15 = ("Apple", "Banana", "Mango")
print(type(t15))
```

```python
# set

set1 = {1, 2, 3}
print(type(set1))

set2 = {"Python", "Java", "C++"}
print(type(set2))

set3 = {10, 20, 30}
print(type(set3))

set4 = set()
print(type(set4))

set5 = {1, "Python", 3.14}
print(type(set5))

set6 = {True, False}
print(type(set6))

set7 = {"Data", "Science"}
print(type(set7))

set8 = {100}
print(type(set8))

set9 = {1, 2, 2, 3, 3}
print(type(set9))

set10 = {10.5, 20.5, 30.5}
print(type(set10))

set11 = {"A", "B", "C"}
print(type(set11))

set12 = {True, 10, "Python"}
print(type(set12))

set13 = {1, 5, 10, 15}
print(type(set13))

set14 = {-1, -2, -3}
print(type(set14))

set15 = {"Apple", "Banana", "Mango"}
print(type(set15))
```

```python
# dict

d1 = {"name": "Jitendra"}
print(type(d1))

d2 = {"name": "Jitendra", "age": 30}
print(type(d2))

d3 = {"city": "Pune", "country": "India"}
print(type(d3))

d4 = {}
print(type(d4))

d5 = {1: "Python", 2: "Java"}
print(type(d5))

d6 = {"a": 10, "b": 20}
print(type(d6))

d7 = {"name": "John", "salary": 50000}
print(type(d7))

d8 = {"Python": "Programming Language"}
print(type(d8))

d9 = {"id": 101, "name": "Alex", "active": True}
print(type(d9))

d10 = {"numbers": [1, 2, 3]}
print(type(d10))

d11 = {"person": {"name": "John", "age": 25}}
print(type(d11))

d12 = {"a": 1, "b": 2, "c": 3}
print(type(d12))

d13 = {1: 100, 2: 200, 3: 300}
print(type(d13))

d14 = {"course": "Python", "duration": "3 months"}
print(type(d14))

d15 = {"name": "Jitendra", "skills": ["Python", "SQL", "PySpark"]}
print(type(d15))
```
