# Create

```python
students = [

["Rahul", 21, "Python"],

["Priya", 22, "Data Science"],

["Aman", 20, "Machine Learning"]

]
```

# Without using loops

```python
# Print the complete list

print(students)

# Print Rahul's name

print(students[0][0])

# Print Priya's age

print(students[1][1])

# Print Aman's course

print(students[2][2])

# Print the complete record of Priya

print(students[1])

# Change Rahul's course to "AI"

students[0][2]="AI"

# Add another student record manually

students.append(["Jitendra",33,"AI"])

# Print the updated nested list

print(students)
```

# Explain how indexing works with nested lists.

In nested lists, each inside list considered as an element and indexing starts from zeroth Index
from first element till last element by incrementing one. Now in each element, indexing 
works similar. To fetch any value from nested loop, we first identified where is that value
after that we give the index to fetch that value, 1st index value shows the position of list 
inside nested list and 2nd index value show the position of value in the inside list.

example: In students[0][2] first index [0] shows the position of first inside list and
index [2] shows the 2nd element of that inside list.
