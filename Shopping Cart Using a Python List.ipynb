# Create

```python
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]
```

# Perform operations to

```python
# Display all products

print(cart[0:])

# Access first and last products

print(cart[0])
print(cart[-1])

# Add "Webcam"

cart.append("Webcam")

# Insert "USB Hub" at index 2

cart.insert(2,"USB Hub")

# Remove "Mouse"

cart.remove("Mouse")

# Remove the last item using pop()

cart.pop()

# Find the index of "Monitor"

cart.index("Monitor")

# Count occurrences of "Laptop"

cart.count("Laptop")

# Create a copy of the cart

import copy

copy_cart=cart.copy()

# Reverse the cart

cart[::-1]

# Sort the products alphabetically

cart.sort()
```

# Explain the difference between

append(),extend(),insert(),remove(),pop(),clear(),copy(),sort(),reverse()

append(): append the given value at last index
extend(): first unwrap the iterable, and then append at end
insert(): insert the value at given index
remove(): remove first occurance of the given value
pop(): remove the last element
clear(): clear the list
copy(): copy the list into another variable
sort(): sort the value based on key
reverse(): reverse the list
