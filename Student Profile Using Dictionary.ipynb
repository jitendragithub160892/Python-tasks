# Create

```python
student = {

"name": "Rahul",

"age": 22,

"course": "Python",

"city": "Bangalore",

"marks": 88

}
```

# Perform

```python
# Print the complete dictionary

print(student)

# Print the student's name

print(student['name'])

# Print their course

print(student['course'])

# Print all keys

print(student.keys())

# Print all values

print(student.values())

# Print all key-value pairs

print(student.items())

# Change marks from 88 to 92

student['marks']=92
print(student)

# Add "email"

student['email']='jitu123@gmail.com'
print(student)

# Add "phone"

student['phone']=9235239642
print(student)

# Remove "city"

student.pop('city')
print(student)

# Use get() to retrieve "name"

print(student.get("name"))

# Create a copy of the dictionary

student_copy=student.copy()
print(student_copy)
```

# Explain

keys(): extract the keys of dictionary
values(): extract the values of dictionary
items(): extract the keys-values of dictionary
get(): Used to fetch the value of given keys
update(): To update the value of given keys
pop(): To delete the value of given keys
copy(): Copy one dictionary to another dictionary
