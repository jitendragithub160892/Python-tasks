# Create data for three products using a list of dictionaries.

Example structure

```python
products = [

{

"name": "Laptop",

"price": 70000,

"brand": "Dell"

},

{

"name": "Phone",

"price": 40000,

"brand": "Samsung"

},

{

"name": "Tablet",

"price": 30000,

"brand": "Apple"

}

]

```

```python
products
```

# Without loops

```python
# Print all products

print(products)

# Print first product

print(products[0])

# Print second product's price

print(products[1]['price'])

# Print third product's brand

print(products[2]['brand'])

# Change first product's price

products[0]['price']=80000
print(products)

# Add "rating" to the second product

products[1]['rating']="5 star"

# Add another product manually

products.append({'name': 'Camera', 'price': 20000, 'brand': 'Apple'})


# Print the final dataset

print(products)
```

# Explain the combination of

```python
# List: 

students = ["Rahul", "Aman", "Priya"]
print(students)

# Dictionary

student = {"name": "Rahul", "age": 25}
print(student)

# Indexing

print(students[0])

# Keys

print(student.keys())

# Values

print(student.values())
```
