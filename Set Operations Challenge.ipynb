# Create

```python
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}

java_students = {"Priya", "Karan", "Rohit", "Simran"}
```

# Perform

```python
# Print both sets
print(python_students)
print(java_students)

# Find students learning either Python or Java

java_or_python_students=python_students.union(java_students)
print(java_or_python_students)

# Find students learning both

both_learning_students=python_students.intersection(java_students)
print(both_learning_students)

# Find students learning only Python

only_python=python_students.difference(java_students)

# Find students learning only Java

only_java=java_students.difference(python_students)

# Find students belonging to exactly one group

exactly_one_group=python_students.symmetric_difference(java_students)

# Add a new student

python_students.add("Jitu")
print(python_students)

# Remove a student

java_students.remove("Priya")
print(java_students)

# Demonstrate discard()

python_students2=python_students.discard("Rahul")
print(python_students)
```

# Use and explain

union(): Combine both sets unique elements
intersection(): Extract both sets common elements
difference(): give only one of the group
symmetric_difference(): Exactly one of the group
add(): add new elements to a set
remove(): remove an existing elements from the set. If element is not present the it raises a exception
discard(): remove element from the set if element is not present then it don't raise any exception.
