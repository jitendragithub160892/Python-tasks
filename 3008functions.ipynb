```python
def factorial(n):
    if n==0 or n==1:
        return 1
    return n* factorial(n-1)
```

```python
factorial(5)
```

```python
def total(n):
    if n==1:
        return 1
    return n+total(n-1)
```

```python
total(5)
```

```python
def calculate_time(func):
    def wrapper(*args,**kwargs):
        start=time.time()
        result=func(*args,**kwargs)
        end=time.time()
        print("total time execution is ",end-start)
        return result
    return wrapper
```

```python
def add(a,b):
    time.sleep(2)
    c=a+b
    return c
```

```python
add(3,4)
```

```python
@calculate_time
def add(a,b):
    time.sleep(2)
    c=a+b
    return c
```

```python
add(3,4)
```
