# Create a dictionary representing a laptop

```python
laptop = {

"brand": "Dell",

"model": "XPS 15",

"price": 120000,

"ram": "16GB",

"storage": "512GB SSD",

"available": True

}
```

# Perform operations to

```python
# Display the product

print(laptop)

# Print brand

print(laptop['brand'])

# Print model

print(laptop['model'])

# Print price

print(laptop['price'])

# Change price

laptop['price']=200000
print(laptop)

# Add processor information

laptop['processor']='Apple'
print(laptop)

# Add GPU information

laptop['GPU']='Nvidia GeForce RTX 50'
print(laptop)

# Change RAM to "32GB"

laptop['ram']="32GB"
print(laptop)

# Remove "available"

laptop.pop('available')
print(laptop)

# Print all keys

print(laptop.keys())

# Print all values

print(laptop.values())

# Print all items

print(laptop.items())

# Finally, create a second product dictionary for a mobile phone.

mobile_phone = {
    "name": "Samsung Galaxy S25",
    "brand": "Samsung",
    "price": 79999,
    "storage": "256GB",
    "ram": "12GB",
    "color": "Black"
}

print(mobile_phone)
```
